# Task 1 — MNIST Classification (Winstars Internship Test 2025)

This task demonstrates three different approaches to classify handwritten digits from the MNIST dataset.

Necessary imports

```python
from data.load_data import load_mnist
from mnist_classifier import MnistClassifier
```

Loading MNIST dataset

```python
X_train, X_test, y_train, y_test = load_mnist()
```

Random Forest model training and evaluation

```python
clf_rf = MnistClassifier(algorithm='rf')
clf_rf.train(X_train, y_train)
```

```python
clf_rf.predict(X_test, y_test);
```

Feed forward NN model training and evaluation

```python
clf_nn = MnistClassifier(algorithm='nn', epochs=10)
clf_nn.train(X_train, y_train)
```

```python
clf_nn.predict(X_test, y_test);
```

Convolutional NN model training and evaluation

```python
clf_cnn = MnistClassifier(algorithm='cnn', epochs=10)
clf_cnn.train(X_train, y_train)
```

```python
clf_cnn.predict(X_test, y_test);
```
